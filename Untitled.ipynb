# BackTrader Strategy Tester

```python
import backtrader as bt
import ccxt
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

# ───────── PARAMETERS ─────────────────────────────
SYMBOL           = 'ALCH/USDT'
TIMEFRAME        = '15m'
HIST_LIMIT       = 2000

ATR_PERIOD       = 14
TP_MULTIPLIER    = 2.0
SL_MULTIPLIER    = 1.0
ERROR_THRESHOLD = 0.01
RISK_AMOUNT      = 10.0  # USD risk per trade

# ───────── FETCH & PREPARE DATA ───────────────────
exchange = ccxt.binance({'enableRateLimit': True})
ohlcv    = exchange.fetch_ohlcv(SYMBOL, TIMEFRAME, limit=HIST_LIMIT)
df       = pd.DataFrame(ohlcv, columns=['ts','open','high','low','close','volume'])
df['datetime'] = pd.to_datetime(df['ts'], unit='ms')
df.set_index('datetime', inplace=True)
df.drop(columns=['ts'], inplace=True)

# Compute features & AD line
df['price_change']  = df['close'].pct_change()
mfm = ((df['close'] - df['low']) - (df['high'] - df['close'])) / (df['high'] - df['low'])
df['mfv']           = mfm * df['volume']
df['ad_line']       = df['mfv'].cumsum()
df['ad_change']     = df['ad_line'].pct_change()

# ATR via pandas
hl  = df['high'] - df['low']
hc  = (df['high'] - df['close'].shift()).abs()
lc  = (df['low']  - df['close'].shift()).abs()
tr  = pd.concat([hl, hc, lc], axis=1).max(axis=1)
df['atr'] = tr.rolling(ATR_PERIOD, min_periods=ATR_PERIOD).mean()

df['volume_change'] = df['volume'].pct_change()

# Drop NaNs for training
feat_cols = ['price_change','ad_change','atr','volume_change']
df_feats  = df[feat_cols].dropna()

# ───────── TRAIN AUTOENCODER ─────────────────────
scaler   = StandardScaler()
X_scaled = scaler.fit_transform(df_feats.values)
model = Sequential([
    Input(shape=(4,)),
    Dense(16, activation='relu'),
    Dense(8, activation='relu'),
    Dense(4, activation='relu'),
    Dense(2, activation='relu', name='bottleneck'),
    Dense(4, activation='linear', name='decoder_output'),
])
model.compile(optimizer='adam', loss='mse')
model.fit(X_scaled, X_scaled, epochs=50, verbose=0)

# Dynamically set threshold based on training reconstruction errors (90th percentile)
reconstructions = model.predict(X_scaled, verbose=0)
train_errors    = np.mean((X_scaled - reconstructions)**2, axis=1)
# ERROR_THRESHOLD = np.percentile(train_errors, 90)
print(f'Dynamic ERROR_THRESHOLD set to: {ERROR_THRESHOLD:.4f}')

# ───────── CUSTOM DATA FEED ──────────────────────
class PandasDataWithAD(bt.feeds.PandasData):
    lines = ('ad_line',)
    params = (('ad_line', 'ad_line'),)

# ───────── BACKTEST STRATEGY ─────────────────────
class ADAutoencoderStrategy(bt.Strategy):
    params = dict(
        error_thr=ERROR_THRESHOLD,
        tp_mul=TP_MULTIPLIER,
        sl_mul=SL_MULTIPLIER,
        risk_amt=RISK_AMOUNT,
        scaler=scaler,
        ae=model,
    )

    def log(self, txt, dt=None):
        dt = dt or self.datas[0].datetime.datetime(0)
        print(f'{dt.isoformat()} | {txt}')

    def __init__(self):
        self.ad_line = self.datas[0].ad_line
        self.atr     = bt.indicators.ATR(self.datas[0], period=ATR_PERIOD)
        self.pchg    = bt.indicators.PercentChange(self.datas[0].close, period=1)
        self.vchg    = bt.indicators.PercentChange(self.datas[0].volume, period=1)
        self.adchg   = bt.indicators.PercentChange(self.ad_line, period=1)

    def notify_trade(self, trade):
        if trade.isclosed:
            self.log(f'TRADE CLOSED | Gross PnL: {trade.pnl:.2f} | Net PnL: {trade.pnlcomm:.2f}')

    def next(self):
        feats = [
            float(self.pchg[0]),
            float(self.adchg[0]),
            float(self.atr[0]),
            float(self.vchg[0])
        ]
        if any(np.isnan(feats)):
            return

        x_sc   = self.p.scaler.transform(np.array(feats).reshape(1, -1))
        x_pred = self.p.ae.predict(x_sc, verbose=0)
        error  = float(np.mean((x_sc - x_pred)**2))

        ad_now, ad_prev = float(self.ad_line[0]), float(self.ad_line[-1])
        direction = 'neutral'
        if ad_now > ad_prev and error > self.p.error_thr:
            direction = 'bullish'
        elif ad_now < ad_prev and error > self.p.error_thr:
            direction = 'bearish'

        price = self.datas[0].close[0]
        sl = self.atr[0] * self.p.sl_mul
        tp = self.atr[0] * self.p.tp_mul

        if direction == 'bullish' and not self.position:
            size = self.p.risk_amt / sl
            ad_dir_symbol = '↑'
            self.log(f'ERROR={error:.4f} THR={self.p.error_thr:.4f} AD_DIR={ad_dir_symbol} SIGNAL={direction}')
            self.log(f'BUY CREATE | Price: {price:.4f} | Size: {size:.4f}')
            self.buy(size=size)
            self.sell(exectype=bt.Order.Stop, price=price - sl, size=size)
            self.sell(exectype=bt.Order.Limit, price=price + tp, size=size)

        elif direction == 'bearish' and not self.position:
            size = self.p.risk_amt / sl
            ad_dir_symbol = '↓'
            self.log(f'ERROR={error:.4f} THR={self.p.error_thr:.4f} AD_DIR={ad_dir_symbol} SIGNAL={direction}')
            self.log(f'SELL CREATE | Price: {price:.4f} | Size: {size:.4f}')
            self.sell(size=size)
            self.buy(exectype=bt.Order.Stop, price=price + sl, size=size)
            self.buy(exectype=bt.Order.Limit, price=price - tp, size=size)

# ───────── RUN CEREBRO ──────────────────────────
if __name__ == '__main__':
    cerebro = bt.Cerebro()
    cerebro.adddata(PandasDataWithAD(dataname=df))
    cerebro.addstrategy(ADAutoencoderStrategy)

    cerebro.addanalyzer(bt.analyzers.TradeAnalyzer, _name='trade_analyzer')
    cerebro.broker.setcash(10000)
    cerebro.broker.setcommission(commission=0.0002)

    print('Start Value:', cerebro.broker.getvalue())
    results = cerebro.run()
    strat = results[0]
    print('Final Value:', cerebro.broker.getvalue())

    trade_analysis = strat.analyzers.trade_analyzer.get_analysis()
    print('\n===== Trade Analysis =====')
    print(f"Total Trades: {trade_analysis.total.closed}")
    print(f"Won: {trade_analysis.won.total} | Lost: {trade_analysis.lost.total}")
    try:
        gross = trade_analysis.pnl.gross.total
        net   = trade_analysis.pnl.net.total
        print(f"PnL Gross: {gross:.2f} | PnL Net: {net:.2f}")
    except Exception:
        print(f"PnL details: {trade_analysis.pnl}")

    # Uncomment to plot
    # cerebro.plot()
```
