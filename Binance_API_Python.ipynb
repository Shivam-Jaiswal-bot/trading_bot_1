```python
# 3.1.1  Get Binance Account Info
account_info = client.get_account()

balances = {
    b['asset']: float(b['free']) + float(b['locked'])
    for b in account_info['balances']
    if float(b['free']) + float(b['locked']) > 0
}

print("Your balances:")
for asset, amt in balances.items():
    print(f"  {asset}: {amt}")
```
