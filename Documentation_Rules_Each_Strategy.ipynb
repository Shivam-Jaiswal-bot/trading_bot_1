### 3.2.2 Position Sizing
- • Quantity = Risk Amount / Price, Risk Amount = $10.
  
- • 5x leverage: $10 trade controls $50 of ALCH/USDT.
  
- • Example: Price = $100, Quantity = 10 / 100 = 0.1 units, Effective Size = 0.1 ×100 × 5 = $50.


### 3.2.3 Take-Profit and Stop-Loss

- • Long TP: Entry Price + 2 × ATR.

- • Long SL: Entry Price - 1 x ATR.

- • Short TP: Entry Price - 2 x ATR.

- • Short SL: Entry Price + 1 x ATR.

- • Example: Entry = $100, ATR = $2, Long TP = $104, SL = $98.
